## 3.Robust Scaling

```python
'''Formula: x_scaled = (x - median) / IQR

যেখানে:
  x       = original value
  median  = feature-এর মধ্যবর্তী value (50th percentile)
  IQR     = Q3 - Q1 (75th percentile - 25th percentile)
  
  '''

'''
Why Median and IQR?
-- Outlier thakle Mean huge rokomer change hoye jay, but Median reamin 
unchanged. Beacuse Median suhdu middle er value dekhe, extreme values na.

Outlier thakleo Median and IQR stable thake. That's why Robust Scaling Outlier 
thakleo baki data ke correctly scale korte pare.

When Robust scaling will be used?
-- jokhon Data te outlier thake and sei outlier remove or clip kora possible na

'''
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../DataSets_File/EDA_file/marks_dataset.csv")
```

```python
df.sample(5)
```

```python
sns.kdeplot(data=df, x = 'maths_marks')
```

```python
df.describe()
```

```python
df['maths_marks'].describe()
```

```python
from sklearn.preprocessing import RobustScaler

rs = RobustScaler().set_output(transform='pandas')

rs.fit(df[['maths_marks']])
df['maths_marks'] = rs.transform(df[['maths_marks']])
```
