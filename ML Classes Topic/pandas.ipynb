## Transformation to df from in built datas

```python
my_list = [['Alice',25],['Bob',30],['Chalie',28]]
list_df = pd.DataFrame(my_list, columns=['Name','Age'],index=['a','b','c'])
print(list_df)
print(type(list_df))
```

```python
my_tuple = (('Alice',25),('Bob',30),('Chalie',28))
tuple_df = pd.DataFrame(my_tuple, columns=['Name','Age'],index=[1,2,3])
print(tuple_df)
print(type(tuple_df))
```

```python
my_dict ={
    'Name': ['Alice','Bob','Charlie'],
    'Age': [25,30,28]
}

dict_df = pd.DataFrame(my_dict, index=[1,2,3])
print(dict_df)
```

```python
my_data = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 28, 'City': 'Dhaka'},
    {'Name': 'Mahin', 'Age': 29}
]
print(my_data)
print(type("my_data : ", my_data))

my_data_df = pd.DataFrame(my_data)
print("\nAfter converting into dataframe\n")
print(my_data_df)
```
