# Outlier Ditection

## 1.Using z-score

### Age

```python
mean_of_age = X_train['Age'].mean()
std_of_age = X_train['Age'].std()

X_train['zscore_age'] = (X_train['Age'] - mean_of_age)/ std_of_age
X_train
```

```python
outliers_age = X_train[abs(X_train['zscore_age'])>3] #Z-score method-এ সাধারণত |Z| > 3 হলে 
#data point-কে outlier ধরা হয়, কারণ normal distribution-এর প্রায় 99.7% data ±3 standard deviation-এর মধ্যে থাকে।

print(len(outliers_age))
display(outliers_age)
```

### Fare

```python
mean_of_fare = X_train["Fare"].mean()
std_of_fare = X_train['Fare'].std()


X_train['zscore_fare'] = (X_train['Fare']-mean_of_fare)/std_of_fare

X_train
```

```python
outliers_fare = X_train[abs(X_train['zscore_fare']) >3]
print(len(outliers_fare))
display(outliers_fare)
```

# 

## 2.IQR(Interquartile range)

```python
# The IQR represents the range of values that lie between Q1 (25th percentile) and 
# Q3 (75th percentile) capturing the middle 50% of the data
```

### Age

```python

age_Q1 = X_train['Age'].quantile(0.25)
age_Q2 = X_train['Age'].quantile(0.5)
age_Q3 = X_train['Age'].quantile(0.75)

age_IQR = age_Q3 - age_Q1

age_minimum = age_Q1 - 1.5 * age_IQR
age_maximum = age_Q3 + 1.5 * age_IQR

print(f"age_minimum: {age_minimum}, age_maximum: {age_maximum}")
# minimum value hote choto value and maximum value hote boro value hocche outliers
```

```python
age_outliers_IQR = X_train[(X_train['Age']<age_minimum) | (X_train['Age']>age_maximum)]
# age_outliers hobe -- Age column e jegulo age_minimum theke choto OR age_maximum theke boro
print(len(age_outliers_IQR))
display(age_outliers_IQR)
```

### Fare

```python
fare_Q1 = X_train['Fare'].quantile(0.25)
fare_Q2 = X_train['Fare'].quantile(0.5)
fare_Q3 = X_train['Fare'].quantile(0.75)

fare_IQR = fare_Q3 - fare_Q1

fare_min = max(0,fare_Q1 - 1.5*fare_IQR)
fare_max = fare_Q3 + 1.5*fare_IQR

print(f"fare minimum: {fare_min}, fare maximum: {fare_max}")
```

```python
fare_outliers_IQR = X_train[(X_train['Fare']<fare_min) | (X_train['Fare']>fare_max)]

print(len(fare_outliers_IQR))
display(fare_outliers_IQR)
```

```python
# z score outlier detection is better for age outliers

X_train = X_train[abs(X_train['zscore_age']) <= 3]
X_train
```

```python
X_train['Fare'] = X_train['Fare'].clip(fare_min, fare_max) #এই code-এর উদ্দেশ্য হলো outlier delete না করে তাদের সীমার মধ্যে এনে রাখা।

print(X_train['Fare'].min())
```

# Scaling

##### ML-এ scaling করা হয় যখন বিভিন্ন feature-এর value/range অনেক আলাদা থাকে।
কেন scaling করা হয়?
যাতে কোনো feature শুধু তার বড় numerical value-এর কারণে model-এর ওপর বেশি influence না করে এবং কিছু algorithm দ্রুত/ভালোভাবে কাজ করে।

Age  → 18–80
Fare → 7–500
এখানে Fare-এর value অনেক বড়। Scaling করলে দুটো feature-কে কাছাকাছি range-এ আনা হয়।

কখন scaling করা হয়?

বিশেষ করে যেসব algorithm distance বা gradient-এর ওপর নির্ভর করে, যেমন:

KNN
K-Means
SVM
Logistic Regression
Linear Regression
Neural Network

⚠️ সবসময় scaling দরকার হয় না। Decision Tree, Random Forest-এর মতো tree-based model-এ সাধারণত scaling প্রয়োজন হয় না।

## 1. Standard Scaling

```python
# Outlier thakle standard scaling kora uchit noy. Karon standard Scaling 
# mean & std use kore, ar ei 2tai outlier sensitive.

'''Standard Scaling data er mean 0 & std 1 e niye ashe'''

'''Formula: z = (x-mean)/std '''

'''Without outlier:  mean = 30,   std = 5.5
   With outlier:     mean = 51,   std = 58.3

ei distorted mean and std diye scale korle normal values gulo sob 0 er kache 
jome jabe. Tokhon model ei values er moddhe difference bujhte parbe na.
'''
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

ss.fit(X_train[['Age']])
X_train['Age'] = ss.transform(X_train[['Age']])
X_test['Age'] = ss.transform(X_test[['Age']])
```
